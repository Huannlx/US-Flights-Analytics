```python
import pandas as pd
import numpy as np
import os
import glob
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)
pd.set_option('display.max_columns',500)
pd.set_option('display.max_rows',500)
pd.set_option('display.width',1000)
```

```python
folder_path = r'D:\T100_SEGMENT_US'
file_format = os.path.join(folder_path,"*_T100_SEGMENT_US_CARRIER_ONLY.csv")
file_path = glob.glob(file_format)
dfs = []
for file in file_path:
    print(file)
    read_file = pd.read_csv(file,low_memory=False)
    dfs.append(read_file)
if dfs:
    df = pd.concat(dfs,ignore_index=True)
    print(df.head(3))
```

```python
df.isna().sum()
```

```python
df[df['UNIQUE_CARRIER'].isna()]
```

```python
df.dropna(subset='UNIQUE_CARRIER',inplace=True)
df.isna().sum()
```

```python
df[df['ORIGIN_CITY_NAME'].isna()]
```

```python
df.loc[(df['ORIGIN'] == 'M91'),'ORIGIN_CITY_NAME'] = 'Springfield, TN'
df.isna().sum()
```

```python
df = df[df['DEPARTURES_PERFORMED'] > 0]
df.shape
```

```python
df = df[~((df['SEATS'] == 0) & (df['PASSENGERS'] > 0))]
df.shape
```

```python
df = df[~((df['AIR_TIME'] == 0) & (df['DISTANCE'] > 0))]
df.shape
```

```python
df = df[~((df['DISTANCE'] == 0) & (df['AIR_TIME'] > 0))]
df.shape
```

```python
df['YEAR_MONTH'] = df['YEAR'].astype(str) + df['MONTH'].astype(str).str.zfill(2)
df.head(3)
```

```python
df['Route_flight'] = df['ORIGIN'].astype(str) + ' - ' + df['DEST'].astype(str)
df.head(3)
```

```python
df = df[~(df['AIR_TIME']> df['RAMP_TO_RAMP'])]
df.shape
```

```python
Fact_US_Carrier = df[['DEPARTURES_SCHEDULED','DEPARTURES_PERFORMED','PAYLOAD','SEATS','PASSENGERS','FREIGHT','MAIL','DISTANCE','RAMP_TO_RAMP','AIR_TIME','UNIQUE_CARRIER','CARRIER_GROUP','ORIGIN_AIRPORT_ID','DEST_AIRPORT_ID','AIRCRAFT_TYPE','AIRCRAFT_GROUP','YEAR_MONTH','Route_flight']]
Fact_US_Carrier.columns = ['Departures_Schedule','Departures_Performed','Payload','Seats','Passengers','Freight','Mail','Distance','Ramp_to_Ramp','Air_Time','Carrier_ID','Carrier_Group_ID','Origin_Airport_ID','Dest_Airport_ID','Aircraft_Type','Aircraft_Group','Date_ID','Route_flight']
Fact_US_Carrier.head(5)
```

```python
Dim_Date = df[['YEAR_MONTH','YEAR','MONTH']].drop_duplicates().sort_values(by=['YEAR', 'MONTH'], ascending=[True, True]).reset_index(drop=True)
Dim_Date.columns = ['Date_ID','Year','Month_num']
Dim_Date.head(5)
```

```python
month_map = {
    1: 'Jan', 2: 'Feb', 3: 'Mar', 4: 'Apr',
    5: 'May', 6: 'Jun', 7: 'Jul', 8: 'Aug',
    9: 'Sep', 10: 'Oct', 11: 'Nov', 12: 'Dec'
}

Dim_Date['Month'] = Dim_Date['Month_num'].map(month_map)
Dim_Date.head(12)
```

```python
Dim_Carrier = df[['UNIQUE_CARRIER','UNIQUE_CARRIER_NAME']].drop_duplicates().sort_values(by='UNIQUE_CARRIER_NAME', ascending=True).reset_index(drop=True)
Dim_Carrier.columns = ['Carrier_ID','Carrier_Name']
Dim_Carrier.head(5)
```

```python
origin = df[['ORIGIN_AIRPORT_ID','ORIGIN','ORIGIN_CITY_NAME','ORIGIN_STATE_ABR','ORIGIN_STATE_NM','ORIGIN_COUNTRY','ORIGIN_COUNTRY_NAME']]\
            .drop_duplicates(subset='ORIGIN_AIRPORT_ID')\
            .sort_values(by='ORIGIN_AIRPORT_ID', ascending=True)\
            .reset_index(drop=True)
origin.columns = ['Airport_ID','Airport_code','City_Name','State_ABR','State_Name','Country_ABR','Country']
origin.head(5)
```

```python
dest = df[['DEST_AIRPORT_ID','DEST','DEST_CITY_NAME','DEST_STATE_ABR','DEST_STATE_NM','DEST_COUNTRY','DEST_COUNTRY_NAME']] \
            .drop_duplicates(subset='DEST_AIRPORT_ID') \
            .sort_values(by='DEST_AIRPORT_ID', ascending=True) \
            .reset_index(drop=True)
dest.columns = ['Airport_ID','Airport_code','City_Name','State_ABR','State_Name','Country_ABR','Country']
dest.head(5)
```

```python
Dim_Airport = pd.concat([origin,dest],axis=0, ignore_index=True)
Dim_Airport.head(5)
```

```python
Dim_Airport = Dim_Airport.drop_duplicates(subset='Airport_ID').reset_index(drop=True)
Dim_Airport.shape
```
